## Ejercicio 2: Escalamiento de tickets de soporte técnico

En este ejercicio se diseña un agente encargado de evaluar tickets de soporte técnico considerando el tiempo de espera, el nivel de urgencia y si el cliente es premium.

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **P - Medida de desempeño** | Atender y escalar correctamente los tickets según su prioridad, reduciendo los tiempos de espera y evitando retrasos en casos importantes. |
| **E - Entorno** | Sistema de soporte técnico, clientes, tickets y equipo de atención. |
| **A - Acciones** | Mantener en cola, priorizar o escalar el ticket. |
| **S - Percepción** | Tiempo de espera en minutos, nivel de urgencia y condición de cliente premium. |
| **Objetivo** | Determinar la prioridad adecuada de cada ticket para brindar una atención eficiente. |

### Justificación de las reglas

Las reglas consideran principalmente el nivel de urgencia del ticket, pero también toman en cuenta el tiempo de espera y si el cliente es premium. Los casos de urgencia alta requieren atención inmediata, mientras que los casos de urgencia media o baja pueden aumentar su prioridad cuando llevan demasiado tiempo esperando. Además, ser cliente premium permite priorizar algunos tickets para brindar una atención más rápida.

#### Reglas de decisión

1. Si la urgencia es alta, el ticket se escala inmediatamente.
2. Si la urgencia es media y el tiempo de espera es mayor o igual a 30 minutos, se escala el ticket.
3. Si la urgencia es media y el cliente es premium, se prioriza el ticket.
4. Si la urgencia es baja, pero el cliente es premium y lleva 60 minutos o más esperando, se escala el ticket.
5. Si la urgencia es baja y lleva 60 minutos o más esperando, se prioriza el ticket.
6. En los demás casos, el ticket se mantiene en cola.

### Código

A continuación se implementa el agente de soporte utilizando las reglas de decisión definidas anteriormente.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    if nivel_urgencia == "alta":
        return (
            "escalar",
            "el ticket presenta un nivel de urgencia alto"
        )

    elif nivel_urgencia == "media":
        if tiempo_espera_minutos >= 30:
            return (
                "escalar",
                "el ticket tiene urgencia media y lleva 30 minutos o más esperando"
            )
        elif cliente_premium:
            return (
                "priorizar",
                "el ticket tiene urgencia media y pertenece a un cliente premium"
            )
        else:
            return (
                "mantener en cola",
                "el ticket tiene urgencia media pero aún no supera el tiempo de espera establecido"
            )

    elif nivel_urgencia == "baja":
        if cliente_premium and tiempo_espera_minutos >= 60:
            return (
                "escalar",
                "el cliente es premium y el ticket lleva 60 minutos o más esperando"
            )
        elif tiempo_espera_minutos >= 60:
            return (
                "priorizar",
                "el ticket lleva 60 minutos o más esperando"
            )
        else:
            return (
                "mantener en cola",
                "el ticket tiene urgencia baja y no requiere atención inmediata"
            )
```

```python
agente_soporte(40, "media", True)
```
